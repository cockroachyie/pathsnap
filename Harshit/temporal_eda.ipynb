# Temporal EDA and Movement Analysis

This notebook analyses GPS timestamps, sampling gaps, trip durations, distance, speed, acceleration, stationary periods, and suspicious movement.

```python
timestamps = pd.read_csv(harshit_dir / "timestamp_summary.csv")
movements = pd.read_csv(harshit_dir / "movement_summary.csv")
display(timestamps.head())
display(movements.head())
```

## Timestamp Findings

```python
print("Trips:", len(timestamps))
print("Average trip duration:", timestamps["duration_seconds"].mean())
print("Median trip duration:", timestamps["duration_seconds"].median())
print("Maximum timestamp gap:", timestamps["maximum_gap_seconds"].max())
```

## Suspicious Trips

```python
display(timestamps.nlargest(10, "maximum_gap_seconds")[["trip", "dataset", "maximum_gap_seconds", "gaps_over_5_seconds"]])
```

```python
display(movements.nlargest(10, "maximum_speed_kph")[["trip", "dataset", "maximum_speed_kph", "speeds_over_180_kph"]])
```

## Conclusion

Most observations use 1-second sampling. Large timestamp gaps can inflate speed and acceleration calculations. Reliable movement results should exclude gaps longer than 5 seconds and speeds outside a reasonable 0-250 km/h range.
