# PathSnap — Model: Random Forest (bagging of decision trees)

**Family:** Ensemble / bagging. **Task framing:** every GPS observation has a set of candidate road arcs
(all arcs within 50 m, at most 20). Each (observation, candidate) pair is one row; the model predicts
`label = 1` if the candidate is the reference arc. The highest-probability candidate is the observation-level
prediction, and the probabilities become **emission scores** for the HMM/Viterbi stage.

**Same data, features, CV and metrics as the other PathSnap model notebooks (Logistic Regression, XGBoost):**
- features and labels from `Feature Extraction/extract_features.py` → `data_features/` (24 features),
- training on the 76 `use_for_tuning` train trips, rows with `label = -1` ignored,
- `GroupKFold(n_splits=5)` on `trip_id` (identical folds to the other notebooks),
- Top-1 counts observations whose reference arc has no candidate as **wrong**,
- student_test (10 trips) is used only to measure finished models — never to choose hyper-parameters.

Extra columns required by the report are also computed: row **Macro F1**, observation Macro F1,
**route continuity**, inference time per observation, and a train-vs-validation **overfitting** check.

## 1. Load data

Same loader as the Logistic Regression / XGBoost notebooks, with two additions:
- rows with `label = -1` are **kept in memory but never used for training or scoring** — they are needed so that
  every observation gets a prediction (for route continuity and the maps);
- each observation gets its `segment_id` (from `track_clean.csv`) and raw lon/lat, for continuity and maps.

```python
flags = pd.read_csv(FLAGS, dtype={"trip_id": str})

def trips_of(part):
    t = flags[flags.partition == part]
    if part == "train":
        t = t[t.use_for_tuning.astype(bool)]
    return sorted(t.trip_id)

def load_partition(part):
    rows, pts = [], []
    for tid in trips_of(part):
        f = pd.read_csv(FEAT_DIR / part / f"{tid}.csv.gz", dtype={"trip_id": str})
        y = pd.read_csv(FEAT_DIR / "labels" / part / f"{tid}.csv.gz")
        assert len(f) == len(y) and (f.idx.values == y.idx.values).all() and (f.arc_id.values == y.arc_id.values).all()
        f["label"] = y.label.values
        f[FEATURES] = f[FEATURES].astype(np.float32)          # halves memory, no effect on trees
        rows.append(f)
        p = pd.read_csv(FEAT_DIR / "labels" / part / f"{tid}_points.csv.gz")
        trk = pd.read_csv(CLEAN_DIR / part / tid / "track_clean.csv", usecols=["idx", "lon", "lat", "segment_id"])
        p = p.merge(trk, on="idx", how="left")
        p.insert(0, "trip_id", tid); p["part"] = part
        pts.append(p)
    return pd.concat(rows, ignore_index=True), pd.concat(pts, ignore_index=True)

t0 = time.time()
train_all, train_pts = load_partition("train")
test_all, test_pts = load_partition("student_test")
train = train_all[train_all.label >= 0].reset_index(drop=True)     # training / scoring rows
print(f"loaded in {time.time() - t0:.1f}s")
for name, d, p in [("train", train, train_pts), ("student_test", test_all[test_all.label >= 0], test_pts)]:
    L = p[p.labelled]
    print(f"{name:13} trips {d.trip_id.nunique():3} | rows {len(d):>9,} | positive rate {d.label.mean():.3f} "
          f"| labelled observations {len(L):>7,} | candidate miss rate {1 - L.ref_in_candidates.mean():.4%}")
```

## 2. Metrics

**Row level** (threshold 0.5, rows with `label ≥ 0`): accuracy, precision / recall / F1 of the true-arc class,
**Macro F1** (mean F1 of both classes), log loss. ~88% of rows are negatives, so row accuracy alone is misleading.

**Observation level** (the primary metrics):
- **Top-1 accuracy** — for each labelled observation, the highest-probability candidate vs the reference arc.
  Observations whose reference arc is not a candidate count as **wrong** (same as the other notebooks).
- **Observation Macro F1** — macro F1 over arc IDs (`trip:arc`) of the Top-1 predictions; rare arcs count equally.
- **Route continuity** — share of consecutive observations (same segment) whose predicted arcs are equal or
  connected within ≤ 2 arcs in the directed network.
- **Nearest-arc rule** (`dist_rank == 0`) — non-ML baseline.

```python
_net = {}
def network(part, tid):
    # arc endpoints + successor sets of each node (canonical arcs) for continuity
    if (part, tid) not in _net:
        z = np.load(CLEAN_DIR / part / tid / "network_clean.npz")
        c = z["arc_is_canonical"]; u, v = z["arc_u"], z["arc_v"]
        _net[(part, tid)] = (u, v, pd.Series(v[c]).groupby(u[c]).apply(set).to_dict())
    return _net[(part, tid)]

def predict_arcs(df, score, pts):
    # Highest-scoring candidate for every observation of the trips in df (-1 = no candidate).
    d = df[["trip_id", "idx", "arc_id"]].assign(s=score)
    best = d.loc[d.groupby(["trip_id", "idx"]).s.idxmax(), ["trip_id", "idx", "arc_id"]]
    P = pts[pts.trip_id.isin(df.trip_id.unique())].merge(best.rename(columns={"arc_id": "pred_arc"}),
                                                         on=["trip_id", "idx"], how="left")
    P["pred_arc"] = P.pred_arc.fillna(-1).astype(int)
    return P

def continuity(P):
    ok = tot = 0
    for (part, tid), g in P.groupby(["part", "trip_id"]):
        u, v, s = network(part, tid)
        g = g.sort_values("idx"); a = g.pred_arc.values; sg = g.segment_id.values
        for i in range(1, len(a)):
            if sg[i] != sg[i - 1] or a[i] < 0 or a[i - 1] < 0:
                continue
            tot += 1; p, q = a[i - 1], a[i]
            ok += (p == q) or (v[p] == u[q]) or (u[q] in s.get(v[p], ()))
    return ok / max(tot, 1)

def obs_metrics(P):
    L = P[P.labelled]
    hit = L.pred_arc == L.ref_arc
    return {"top1": hit.mean(), "top1_given_candidate": hit.sum() / max(L.ref_in_candidates.sum(), 1),
            "obs_macro_f1": f1_score(L.trip_id + ":" + L.ref_arc.astype(str),
                                     L.trip_id + ":" + L.pred_arc.astype(str), average="macro"),
            "n_obs": len(L)}

def evaluate(model, df, pts, with_continuity=False, nearest=False):
    # df may contain label -1 rows: they get predictions but are never scored.
    t0 = time.time(); p = model.predict_proba(df[FEATURES])[:, 1]; pred_s = time.time() - t0
    lab = df.label.values >= 0
    y, pl = df.label.values[lab], p[lab]
    yhat = (pl >= 0.5).astype(int)
    m = {"accuracy": accuracy_score(y, yhat), "precision": precision_score(y, yhat, zero_division=0),
         "recall": recall_score(y, yhat, zero_division=0), "f1": f1_score(y, yhat, zero_division=0),
         "macro_f1": f1_score(y, yhat, average="macro"),
         "log_loss": log_loss(y, np.clip(pl, 1e-7, 1 - 1e-7), labels=[0, 1])}
    P = predict_arcs(df, p, pts)
    m.update(obs_metrics(P))
    m["predict_s"] = pred_s
    m["infer_ms_per_obs"] = 1000 * pred_s / max(P.shape[0], 1)
    if with_continuity:
        m["route_continuity"] = continuity(P)
    if nearest:
        m["top1_nearest_arc"] = obs_metrics(predict_arcs(df, -df.dist_rank.values.astype(float), pts))["top1"]
    return m, p, P
```

## 3. Random Forest

**Why it suits the task**
- The rules are non-linear and interacting: a small distance only helps if the heading agrees; heading only means
  something when the vehicle moves; context distance matters most at junctions. Trees learn such if-then rules
  directly and need no feature scaling.
- A single deep tree has **high variance** (it memorises its training trips). A forest averages many
  de-correlated trees — each fitted on a **bootstrap sample** of rows and choosing every split from a **random subset
  of features** — which lowers variance while keeping the low bias of deep trees.
- Averaging gives **smoother probabilities** than one tree's leaf frequencies, which matters because the
  probabilities become HMM emission scores.

**Limitations:** slower and bigger than a single tree or a linear model; less interpretable; probabilities are
pulled away from 0/1 by averaging (can be under-confident); the out-of-bag estimate is optimistic here because
bootstrap rows from the same trip are correlated, so **group CV is the number we trust**.

**Objective / loss:** every tree is a CART tree grown greedily by minimising **Gini impurity**
$G = 1 - \sum_k p_k^2$ at each split. The forest probability is the mean of the trees' leaf class frequencies.
No global loss is optimised — the variance reduction comes from averaging.

**Cross-validation:** `GroupKFold(5)` on `trip_id` over the label ≥ 0 rows → the same folds as the
Logistic Regression and XGBoost notebooks. For every fold the model is scored on its **own training folds too**,
so the train-vs-validation gap (overfitting) is measured.

**Untuned baseline:** sklearn defaults — `n_estimators=100`, `max_depth=None`, `min_samples_leaf=1`,
`max_features="sqrt"`, `bootstrap=True`.

```python
gkf = GroupKFold(n_splits=5)
folds = list(gkf.split(train, train.label, groups=train.trip_id))

def make_rf(**kw):
    cfg = dict(n_estimators=100, max_features="sqrt", min_samples_leaf=1, bootstrap=True,
               n_jobs=N_JOBS, random_state=SEED)
    cfg.update(kw)
    return RandomForestClassifier(**cfg)

def cv_run(make_model, name, train_scores=True, with_continuity=False):
    rows = []
    for fold, (tr, va) in enumerate(folds):
        a, b = train.iloc[tr], train.iloc[va]
        model = make_model()
        t0 = time.time(); model.fit(a[FEATURES], a.label); fit_s = time.time() - t0
        mv, _, _ = evaluate(model, b, train_pts, with_continuity=with_continuity, nearest=(name == "RF-default"))
        row = {"model": name, "fold": fold, "fit_s": fit_s, **{f"val_{k}": v for k, v in mv.items()}}
        if train_scores:
            mt, _, _ = evaluate(model, a, train_pts)
            row.update({f"train_{k}": v for k, v in mt.items()})
        rows.append(row)
        del model
    return pd.DataFrame(rows)

t0 = time.time()
cv_default = cv_run(lambda: make_rf(), "RF-default", with_continuity=True)
print(f"done in {time.time() - t0:.0f}s")
cv_default[["fold", "val_top1", "val_top1_nearest_arc", "val_macro_f1", "val_f1", "val_log_loss",
            "val_route_continuity", "train_top1", "train_macro_f1", "train_log_loss", "fit_s"]].round(4)
```

## 4. Hyper-parameter sweep (group CV)

A small, explainable grid over the parameters that control how much each tree can memorise
(100 trees per forest during the sweep to keep it affordable; the final model uses 200).
Bayesian (Optuna) tuning is done later only for the model finally selected across all families.

| Hyper-parameter | Values | Effect |
|---|---|---|
| `min_samples_leaf` | 1, 5, 20, 50 | larger = smoother leaves, less memorisation, better probabilities |
| `max_features` | `sqrt` (5 of 24), 0.5 (12 of 24) | features tried per split; fewer = more de-correlated trees |

**Selection rule (decided before looking at results):** highest mean CV **Top-1**; configurations within
0.001 of the best are tie-broken by lower mean CV **log loss** (better HMM emissions).

```python
GRID = [(leaf, mf) for mf in ["sqrt", 0.5] for leaf in [1, 5, 20, 50]]
t0 = time.time(); runs = []
for leaf, mf in GRID:
    r = cv_run(lambda: make_rf(min_samples_leaf=leaf, max_features=mf), f"leaf={leaf}, mf={mf}")
    runs.append(r.assign(min_samples_leaf=leaf, max_features=str(mf)))
    print(f"leaf={leaf:>2} mf={str(mf):>4} | CV top-1 {r.val_top1.mean():.4f} | macro F1 {r.val_macro_f1.mean():.4f} "
          f"| log loss {r.val_log_loss.mean():.4f} | {time.time() - t0:.0f}s")
sweep = pd.concat(runs, ignore_index=True)
sweep.to_csv(OUT / "rf_sweep_folds.csv", index=False)

sweep_s = sweep.groupby(["max_features", "min_samples_leaf"]).agg(
    cv_top1=("val_top1", "mean"), cv_top1_std=("val_top1", "std"), train_top1=("train_top1", "mean"),
    cv_macro_f1=("val_macro_f1", "mean"), train_macro_f1=("train_macro_f1", "mean"),
    cv_log_loss=("val_log_loss", "mean"), train_log_loss=("train_log_loss", "mean"), fit_s=("fit_s", "mean"))
sweep_s.round(4)
```

```python
best = sweep_s.cv_top1.max()
BEST_MF, BEST_LEAF = sweep_s[sweep_s.cv_top1 >= best - 0.001].cv_log_loss.idxmin()
BEST_MF = "sqrt" if BEST_MF == "sqrt" else float(BEST_MF)
BEST_LEAF = int(BEST_LEAF)
BEST_PARAMS = dict(n_estimators=200, min_samples_leaf=BEST_LEAF, max_features=BEST_MF)
SEL = f"RF-tuned (leaf={BEST_LEAF}, mf={BEST_MF})"
print("Selected:", SEL, BEST_PARAMS)

fig, ax = plt.subplots(1, 3, figsize=(17, 4.2))
for mf, g in sweep_s.groupby(level=0):
    g = g.droplevel(0)
    ax[0].plot(g.index, g.cv_top1, marker="o", label=f"CV, mf={mf}")
    ax[0].plot(g.index, g.train_top1, ls="--", marker="s", label=f"train folds, mf={mf}")
    ax[1].plot(g.index, g.cv_macro_f1, marker="o", label=f"CV, mf={mf}")
    ax[1].plot(g.index, g.train_macro_f1, ls="--", marker="s", label=f"train folds, mf={mf}")
    ax[2].plot(g.index, g.cv_log_loss, marker="o", label=f"CV, mf={mf}")
    ax[2].plot(g.index, g.train_log_loss, ls="--", marker="s", label=f"train folds, mf={mf}")
for a, t, yl in zip(ax, ["Top-1 accuracy", "Row Macro F1", "Log loss"], ["Top-1", "Macro F1", "log loss"]):
    a.set(xscale="log", xlabel="min_samples_leaf", ylabel=yl, title=f"{t}: train folds vs CV"); a.grid(alpha=.3); a.legend(fontsize=8)
plt.tight_layout(); plt.savefig(OUT / "rf_validation_curves.png", dpi=140); plt.show()
```

## 5. Number of trees and overfitting analysis

**Trees curve** — with `warm_start`, the selected forest is grown from 10 to 300 trees on fold 0 and scored on the
held-out trips after each step. More trees only reduce variance (they cannot add overfitting); the curve shows where
extra trees stop paying off.

```python
tr, va = folds[0]
a, b = train.iloc[tr], train.iloc[va]
rf_ws = make_rf(min_samples_leaf=BEST_LEAF, max_features=BEST_MF, warm_start=True)
curve = []
for n in [10, 25, 50, 100, 200, 300]:
    rf_ws.set_params(n_estimators=n); rf_ws.fit(a[FEATURES], a.label)
    m, _, _ = evaluate(rf_ws, b, train_pts)
    curve.append({"n_estimators": n, "cv_top1": m["top1"], "cv_macro_f1": m["macro_f1"], "cv_log_loss": m["log_loss"]})
curve = pd.DataFrame(curve).set_index("n_estimators"); del rf_ws

fig, ax = plt.subplots(1, 2, figsize=(12, 3.6))
curve[["cv_top1", "cv_macro_f1"]].plot(ax=ax[0], marker="o"); ax[0].set(title="Fold 0: score vs number of trees")
curve["cv_log_loss"].plot(ax=ax[1], marker="o", c="#8172b2"); ax[1].set(title="Fold 0: log loss vs number of trees")
for x in ax: x.grid(alpha=.3)
plt.tight_layout(); plt.savefig(OUT / "rf_trees_curve.png", dpi=140); plt.show()
curve.round(4)
```

**Train vs validation gap.** Mean over the 5 folds, for the untuned forest and the selected one (200 trees).
A large positive gap in Top-1 / Macro F1 (or a much lower train log loss) means the model memorises its training
trips. Section 7 repeats the check on the full training set vs student_test.

```python
cv_sel = cv_run(lambda: make_rf(**BEST_PARAMS), SEL, with_continuity=True)
cv_all = pd.concat([cv_default, cv_sel], ignore_index=True)
cv_all.to_csv(OUT / "rf_cv_folds.csv", index=False)

of = cv_all.groupby("model", sort=False).agg(
    train_top1=("train_top1", "mean"), cv_top1=("val_top1", "mean"), cv_top1_std=("val_top1", "std"),
    train_macro_f1=("train_macro_f1", "mean"), cv_macro_f1=("val_macro_f1", "mean"), cv_macro_f1_std=("val_macro_f1", "std"),
    train_log_loss=("train_log_loss", "mean"), cv_log_loss=("val_log_loss", "mean"),
    cv_obs_macro_f1=("val_obs_macro_f1", "mean"), cv_continuity=("val_route_continuity", "mean"), fit_s=("fit_s", "mean"))
of["top1_gap (train-CV)"] = of.train_top1 - of.cv_top1
of["macro_f1_gap (train-CV)"] = of.train_macro_f1 - of.cv_macro_f1
of.round(4).T
```

## 6. Final models — fit on all 76 train trips, evaluate once on student_test

Reference models are refit here **on exactly the same rows** so the comparison is like-for-like:
- `DT (depth 8)` — the Decision Tree configuration chosen in the Model 1 notebook (`max_depth=8, min_samples_leaf=20`);
- `Logistic Regression` — the configuration from the GitHub Logistic Regression notebook (StandardScaler + C=1, lbfgs).
  Its student_test Top-1 here should reproduce that notebook's value (0.8616) — a check that both notebooks use the same data.
- `XGBoost` — the configuration from the GitHub XGBoost notebook (hist, 300 trees, depth 10, lr 0.1, subsample 0.8,
  colsample 0.8, no early stopping).

Logistic Regression and XGBoost are also run through the same 5-fold group CV, so every model has a CV column.

```python
def make_logreg():
    return make_pipeline(StandardScaler(), LogisticRegression(C=1.0, solver="lbfgs", max_iter=2000, random_state=SEED))

def make_xgb():
    return XGBClassifier(objective="binary:logistic", eval_metric="logloss", tree_method="hist", n_estimators=300,
                         max_depth=10, learning_rate=0.1, subsample=0.8, colsample_bytree=0.8, min_child_weight=1,
                         reg_lambda=1.0, scale_pos_weight=1.0, n_jobs=-1, random_state=SEED)

t0 = time.time()
for name, make in [("Logistic Regression", make_logreg), ("XGBoost", make_xgb)]:
    cv_all = pd.concat([cv_all, cv_run(make, name, with_continuity=True)], ignore_index=True)
    print(f"{name}: CV done ({time.time() - t0:.0f}s)")
cv_all.to_csv(OUT / "cv_folds_all_models.csv", index=False)

final_models = {
    "RF-default": make_rf(),
    SEL: make_rf(**BEST_PARAMS),
    "DT (depth 8)": DecisionTreeClassifier(max_depth=8, min_samples_leaf=20, random_state=SEED),
    "Logistic Regression": make_logreg(),
    "XGBoost": make_xgb(),
}
results, preds = {}, {}
for name, m in final_models.items():
    t0 = time.time(); m.fit(train[FEATURES], train.label); fit_s = time.time() - t0
    r, p, P = evaluate(m, test_all, test_pts, with_continuity=True, nearest=True)
    cvm = cv_all[cv_all.model == name]
    r.update({"fit_s": fit_s,
              "cv_top1": cvm.val_top1.mean() if len(cvm) else np.nan,
              "cv_macro_f1": cvm.val_macro_f1.mean() if len(cvm) else np.nan,
              "cv_macro_f1_std": cvm.val_macro_f1.std() if len(cvm) else np.nan})
    results[name], preds[name] = r, (p, P)
    print(f"{name:<32} fit {fit_s:6.1f}s | test top-1 {r['top1']:.4f} | macro F1 {r['macro_f1']:.4f} | log loss {r['log_loss']:.4f}")

res = pd.DataFrame(results)
res.to_csv(OUT / "rf_student_test_results.csv")
res.loc[["cv_top1", "cv_macro_f1", "top1", "top1_given_candidate", "top1_nearest_arc", "obs_macro_f1", "macro_f1",
         "f1", "precision", "recall", "accuracy", "log_loss", "route_continuity", "fit_s", "infer_ms_per_obs"]].round(4)
```

## 8. Feature importance

- **Gini (impurity) importance** — total impurity decrease per feature, averaged over trees. Fast but biased
  towards continuous features with many split points.
- **Permutation importance** — increase in student_test log loss when one feature column is shuffled
  (5 repeats, 30,000-row sample). Measured on unseen trips, so it is the more trustworthy of the two.

```python
sub = test.sample(n=min(30000, len(test)), random_state=SEED)
pi = permutation_importance(rf, sub[FEATURES], sub.label, scoring="neg_log_loss", n_repeats=5, random_state=SEED, n_jobs=1)
imp = pd.DataFrame({"gini": rf.feature_importances_, "perm_logloss_increase": pi.importances_mean,
                    "perm_std": pi.importances_std}, index=FEATURES).sort_values("perm_logloss_increase")
imp.to_csv(OUT / "rf_feature_importance.csv")
fig, ax = plt.subplots(1, 2, figsize=(15, 6.5))
imp.gini.sort_values().plot.barh(ax=ax[0], color="#4c72b0"); ax[0].set_title("Gini (impurity) importance")
imp.perm_logloss_increase.plot.barh(ax=ax[1], xerr=imp.perm_std, color="#55a868")
ax[1].set_title("Permutation importance (Δ log loss on student_test sample)")
plt.tight_layout(); plt.savefig(OUT / "rf_feature_importance.png", dpi=140); plt.show()
imp.sort_values("perm_logloss_increase", ascending=False).round(4)
```

## 9. Folium maps — all 10 student_test trips

Basemaps are free and need **no API key**: OpenStreetMap, CartoDB Positron, Esri World Imagery (satellite).
Overlays (toggle in the layer control, top right): road network within 80 m, raw GPS track, reference route (green),
predicted route (red, dashed), GPS observations coloured by correctness (green = correct, red = wrong,
grey = unlabelled), start / end markers. The legend box shows that trip's metrics. Click a point for details.

```python
import folium
from folium.plugins import Fullscreen, MiniMap
from scipy.spatial import cKDTree

def load_net(tid):
    return np.load(CLEAN_DIR / "student_test" / tid / "network_clean.npz")

def arc_line(z, a):
    return [(z["node_lat"][z["arc_u"][a]], z["node_lon"][z["arc_u"][a]]),
            (z["node_lat"][z["arc_v"][a]], z["node_lon"][z["arc_v"][a]])]

def chain(z, arcs):
    # merge consecutive connected arcs into polylines; a break starts a new polyline
    lines, cur = [], []
    for a in arcs:
        seg = arc_line(z, a)
        if cur and cur[-1] == seg[0]:
            cur.append(seg[1])
        else:
            if cur: lines.append(cur)
            cur = list(seg)
    if cur: lines.append(cur)
    return lines

def nearby_arcs(z, g, radius_m=80):
    k = 111320.0; c = np.cos(np.radians(g.lat.mean()))
    d, _ = cKDTree(np.c_[g.lon * k * c, g.lat * k]).query(np.c_[z["node_lon"] * k * c, z["node_lat"] * k],
                                                         distance_upper_bound=radius_m)
    near = np.isfinite(d)
    return np.flatnonzero(z["arc_is_canonical"] & (near[z["arc_u"]] | near[z["arc_v"]]))

def make_map(tid):
    z = load_net(tid); t = per_trip.loc[tid]
    g = P_sel[P_sel.trip_id == tid].sort_values("idx")
    route = pd.read_csv(CLEAN_DIR / "labels" / "student_test" / tid / "route_clean.csv")

    m = folium.Map(tiles=None, control_scale=True)
    folium.TileLayer("openstreetmap", name="OpenStreetMap").add_to(m)
    folium.TileLayer("cartodbpositron", name="CartoDB Positron").add_to(m)
    folium.TileLayer("https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}",
                     attr="Tiles &copy; Esri", name="Esri satellite").add_to(m)

    fg = folium.FeatureGroup("Road network (within 80 m)")
    for a in nearby_arcs(z, g):
        folium.PolyLine(arc_line(z, a), color="#7f7f7f", weight=1, opacity=0.55).add_to(fg)
    fg.add_to(m)

    fg = folium.FeatureGroup("Raw GPS track", show=False)
    for _, s in g.groupby("segment_id"):
        folium.PolyLine(list(zip(s.lat, s.lon)), color="#1f77b4", weight=1.5, opacity=0.7).add_to(fg)
    fg.add_to(m)

    fg = folium.FeatureGroup("Reference route")
    for line in chain(z, route.canonical_arc_id.values):
        folium.PolyLine(line, color="#2ca02c", weight=7, opacity=0.55, tooltip="reference route").add_to(fg)
    fg.add_to(m)

    pa = g.pred_arc[g.pred_arc >= 0].values
    pa = pa[np.r_[True, pa[1:] != pa[:-1]]]                      # drop consecutive duplicates
    fg = folium.FeatureGroup("Predicted route (RF top-1)")
    for line in chain(z, pa):
        folium.PolyLine(line, color="#d62728", weight=3, opacity=0.95, dash_array="6,4", tooltip="predicted").add_to(fg)
    fg.add_to(m)

    fg = folium.FeatureGroup("GPS observations (by correctness)")
    for r in g.itertuples():
        wrong = r.labelled and r.pred_arc != r.ref_arc
        col = "#9e9e9e" if not r.labelled else ("#d73027" if wrong else "#1a9850")
        folium.CircleMarker((r.lat, r.lon), radius=4.5 if wrong else 2.5, color=col, fill=True, fill_opacity=0.9, weight=1,
                            popup=f"idx {r.idx}<br>predicted arc {r.pred_arc}<br>reference arc {r.ref_arc}").add_to(fg)
    fg.add_to(m)

    folium.Marker((g.lat.iloc[0], g.lon.iloc[0]), tooltip="start", icon=folium.Icon(color="green", icon="play")).add_to(m)
    folium.Marker((g.lat.iloc[-1], g.lon.iloc[-1]), tooltip="end", icon=folium.Icon(color="black", icon="stop")).add_to(m)
    folium.LayerControl(collapsed=False).add_to(m)
    Fullscreen().add_to(m); MiniMap(toggle_display=True).add_to(m)

    m.get_root().html.add_child(folium.Element(f'''
    <div style="position:fixed;bottom:24px;left:12px;z-index:9999;background:white;padding:10px 12px;
                border:1px solid #888;border-radius:6px;font:13px Arial;box-shadow:0 1px 4px rgba(0,0,0,.3)">
      <b>PathSnap · trip {tid}</b> · Random Forest (classifier only)<br>
      Top-1 <b>{t.rf_top1:.3f}</b> · obs Macro F1 <b>{t.rf_obs_macro_f1:.3f}</b> · continuity <b>{t.rf_continuity:.3f}</b><br>
      {int(t.labelled_obs):,} labelled obs · {int(t.rf_wrong):,} wrong · {t.mean_candidates:.1f} candidates/obs
      <hr style="margin:5px 0">
      <span style="color:#2ca02c;font-size:16px">&#9473;</span> reference route &nbsp;
      <span style="color:#d62728;font-size:16px">&#9477;</span> predicted route<br>
      <span style="color:#1a9850">&#9679;</span> correct &nbsp;<span style="color:#d73027">&#9679;</span> wrong &nbsp;
      <span style="color:#9e9e9e">&#9679;</span> unlabelled &nbsp;<span style="color:#7f7f7f">&#9472;</span> road network
    </div>'''))
    m.fit_bounds([[g.lat.min(), g.lon.min()], [g.lat.max(), g.lon.max()]], padding=(20, 20))
    return m

for tid in per_trip.index:
    make_map(tid).save(str(MAP_DIR / f"trip_{tid}_rf_map.html"))
print(f"saved {len(per_trip)} maps to {MAP_DIR}")
SHOW_TRIP = per_trip.rf_top1.idxmin()          # hardest trip for the forest — change to any trip id
print("showing trip", SHOW_TRIP)
make_map(SHOW_TRIP)
```

## 11. Viva notes

Quote numbers from the outputs above, not from memory.
1. **Why RF:** bagging + random feature subsets average many de-correlated deep trees → lower variance than one tree,
   smoother probabilities (see log-loss bar and calibration plot).
2. **No test leakage:** hyper-parameters chosen by group CV on the 76 train trips; student_test used once at the end.
3. **Overfitting:** section 5 (train folds vs CV) and section 7 (train vs student_test). Fully grown trees
   (`min_samples_leaf=1`) fit the training trips almost perfectly; larger leaves narrow the gap.
4. **More trees ≠ more overfitting** — trees curve flattens; extra trees only cost time.
5. **Same data as LR/XGBoost:** the Logistic Regression refit here reproduces its own notebook's student_test Top-1.
6. **Errors** concentrate in dense networks (many candidates per observation) and at junctions — where the HMM's
   network-connectivity transitions should help next.
