### Step 7: Calculate Advanced Route Metrics (Top-1 Accuracy, Candidate Miss Rate, & Route Continuity)
We will evaluate the remaining sequence-level metrics using the predictions, candidate sets, and ground-truth sequences from our loaded track files.

```python
def evaluate_advanced_route_metrics(df, predictions):
    # Add predictions back to the data
    df_eval = df.copy()
    df_eval['pred_segment_id'] = predictions

    # 1. Observation-level Top-1 Accuracy
    # Comparing the chosen prediction with ground truth 'segment_id'
    top1_correct = (df_eval['pred_segment_id'] == df_eval['segment_id']).sum()
    top1_accuracy = top1_correct / len(df_eval)

    # 2. Candidate Miss Rate
    # Miss rate is when the true segment_id is missing from candidate set (segment_id == -1 or NaN indicates missing)
    total_observations = len(df_eval)
    missed_candidates = (df_eval['segment_id'] == -1).sum() # Assuming -1 represents missing segments in preprocessed data
    candidate_miss_rate = missed_candidates / total_observations if total_observations > 0 else 0.0

    # 3. Route Continuity (Sequence-level Performance)
    # Compute transitions between sequential nodes. Consecutive segments matching true transitions.
    df_eval['prev_pred'] = df_eval.groupby('source_file')['pred_segment_id'].shift(1)
    df_eval['prev_true'] = df_eval.groupby('source_file')['segment_id'].shift(1)

    # Count coherent continuous transitions
    transitions = df_eval.dropna(subset=['prev_pred', 'prev_true'])
    valid_transitions = ((transitions['pred_segment_id'] == transitions['segment_id']) &
                         (transitions['prev_pred'] == transitions['prev_true'])).sum()
    total_transitions = len(transitions)
    route_continuity = valid_transitions / total_transitions if total_transitions > 0 else 1.0

    print("=" * 45)
    print("       ADVANCED ROUTE METRIC RESULTS        ")
    print("=" * 45)
    print(f"Observation-level Top-1 Accuracy: {top1_accuracy:.5%}")
    print(f"Candidate Miss Rate:              {candidate_miss_rate:.5%}")
    print(f"Route Continuity (Sequence):      {route_continuity:.5%}")
    print("=" * 45)

# Run the analysis on the validation subset
y_val_pred = knn.predict(X_val)
df_val_subset = df_data.iloc[y_val.index]
evaluate_advanced_route_metrics(df_val_subset, y_val_pred)
```

### Step 8: Train KNN on the Complete Cleaned Dataset and Compute All Specified Metrics

```python
import os
import pandas as pd
import numpy as np
import time
import glob
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, log_loss
from sklearn.preprocessing import StandardScaler

clean_dir = '/content/drive/MyDrive/data_clean/'
track_files = glob.glob(os.path.join(clean_dir, 'train/**/track_clean.csv'), recursive=True)
print(f"Compiling all {len(track_files)} cleaned track files...")

all_tracks = []
for file_path in track_files:
    try:
        df_t = pd.read_csv(file_path)
        df_t['source_file'] = os.path.basename(os.path.dirname(file_path))
        all_tracks.append(df_t)
    except Exception as e:
        print(f"Error reading {file_path}: {e}")

df_full = pd.concat(all_tracks, ignore_index=True)
print(f"Successfully compiled full training dataset of shape: {df_full.shape}")

# Set target and features
target_col = 'segment_id'
feature_cols = [c for c in df_full.select_dtypes(include=[np.number]).columns if c != target_col]

X_full = df_full[feature_cols].fillna(0)
y_full = df_full[target_col].fillna(-1)

# Scale full feature set
scaler = StandardScaler()
X_full_scaled = scaler.fit_transform(X_full)

# Perform train-test split (80/20)
X_train, X_val, y_train, y_val = train_test_split(X_full_scaled, y_full, test_size=0.2, random_state=42)

# Train Model and Measure Training Time
knn = KNeighborsClassifier(n_neighbors=5)
t_start_train = time.time()
knn.fit(X_train, y_train)
training_time = time.time() - t_start_train

# Inference and Inference Time
t_start_inf = time.time()
y_pred = knn.predict(X_val)
inference_time = time.time() - t_start_inf

y_prob = knn.predict_proba(X_val)

# Calculate Standard Classification Metrics
accuracy = accuracy_score(y_val, y_pred)
precision = precision_score(y_val, y_pred, average='weighted', zero_division=0)
recall = recall_score(y_val, y_pred, average='weighted', zero_division=0)
f1 = f1_score(y_val, y_pred, average='weighted', zero_division=0)
conf_matrix = confusion_matrix(y_val, y_pred)

try:
    loss = log_loss(y_val, y_prob)
except Exception:
    loss = "N/A"

# Calculate Advanced Sequential and Route Metrics
df_eval = df_full.iloc[y_val.index].copy()
df_eval['pred_segment_id'] = y_pred

# 1. Observation-level Top-1 Accuracy
top1_correct = (df_eval['pred_segment_id'] == df_eval['segment_id']).sum()
top1_accuracy = top1_correct / len(df_eval)

# 2. Candidate Miss Rate
missed_candidates = (df_eval['segment_id'] == -1).sum()
candidate_miss_rate = missed_candidates / len(df_eval)

# 3. Route Continuity / Sequence-level Performance
df_eval['prev_pred'] = df_eval.groupby('source_file')['pred_segment_id'].shift(1)
df_eval['prev_true'] = df_eval.groupby('source_file')['segment_id'].shift(1)
transitions = df_eval.dropna(subset=['prev_pred', 'prev_true'])
valid_transitions = ((transitions['pred_segment_id'] == transitions['segment_id']) &
                     (transitions['prev_pred'] == transitions['prev_true'])).sum()
route_continuity = valid_transitions / len(transitions) if len(transitions) > 0 else 1.0

# Print final scorecard
print("\n" + "="*55)
print("       FULL DATASET KNN PERFORMANCE SCORECARD       ")
print("="*55)
print(f"Training Time:                    {training_time:.4f} seconds")
print(f"Inference Time:                   {inference_time:.4f} seconds")
print(f"Accuracy:                         {accuracy:.5%}")
print(f"Precision:                        {precision:.5%}")
print(f"Recall:                           {recall:.5%}")
print(f"F1-Score:                         {f1:.5%}")
print(f"Log Loss:                         {loss}")
print(f"Observation-level Top-1 Accuracy: {top1_accuracy:.5%}")
print(f"Candidate Miss Rate:              {candidate_miss_rate:.5%}")
print(f"Route Continuity:                 {route_continuity:.5%}")
print("="*55)
print("\nConfusion Matrix:")
print(conf_matrix)
```

### Step 10: Evaluate KNN Using Group-Based Split (True Generalization Validation)
To check if the ~99.9% accuracy was caused by spatial data leakage (where consecutive points of the same trip are split between train and test sets), we will group our split by `source_file`. This ensures the validation set consists of entirely unseen trips.

```python
import os
import pandas as pd
import numpy as np
import time
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GroupKFold

# Group by source_file to perform Group-based splitting
groups = df_full['source_file']

# We will perform a single fold group split (e.g., 5-fold, taking 1 fold as test/validation ~20% of trips)
gkf = GroupKFold(n_splits=5)
train_idx, val_idx = next(gkf.split(X_full, y_full, groups=groups))

# Split features, targets, and groups
X_train_g, X_val_g = X_full.iloc[train_idx], X_full.iloc[val_idx]
y_train_g, y_val_g = y_full.iloc[train_idx], y_full.iloc[val_idx]
df_val_g = df_full.iloc[val_idx].copy()

# Fit scaler ONLY on training data to prevent any scaling leakage
scaler_g = StandardScaler()
X_train_g_scaled = scaler_g.fit_transform(X_train_g)
X_val_g_scaled = scaler_g.transform(X_val_g)

# Train KNN
knn_g = KNeighborsClassifier(n_neighbors=5)
t_start_train = time.time()
knn_g.fit(X_train_g_scaled, y_train_g)
training_time_g = time.time() - t_start_train

# Inference
t_start_inf = time.time()
y_pred_g = knn_g.predict(X_val_g_scaled)
inference_time_g = time.time() - t_start_inf

# Calculate Standard Metrics
accuracy_g = accuracy_score(y_val_g, y_pred_g)
precision_g = precision_score(y_val_g, y_pred_g, average='weighted', zero_division=0)
recall_g = recall_score(y_val_g, y_pred_g, average='weighted', zero_division=0)
f1_g = f1_score(y_val_g, y_pred_g, average='weighted', zero_division=0)
conf_matrix_g = confusion_matrix(y_val_g, y_pred_g)

# Calculate Route Metrics on Group-Split Validation Set
df_val_g['pred_segment_id'] = y_pred_g

top1_correct_g = (df_val_g['pred_segment_id'] == df_val_g['segment_id']).sum()
top1_accuracy_g = top1_correct_g / len(df_val_g)

missed_candidates_g = (df_val_g['segment_id'] == -1).sum()
candidate_miss_rate_g = missed_candidates_g / len(df_val_g)

# Continuity calculation
df_val_g['prev_pred'] = df_val_g.groupby('source_file')['pred_segment_id'].shift(1)
df_val_g['prev_true'] = df_val_g.groupby('source_file')['segment_id'].shift(1)
transitions_g = df_val_g.dropna(subset=['prev_pred', 'prev_true'])
valid_transitions_g = ((transitions_g['pred_segment_id'] == transitions_g['segment_id']) &
                       (transitions_g['prev_pred'] == transitions_g['prev_true'])).sum()
route_continuity_g = valid_transitions_g / len(transitions_g) if len(transitions_g) > 0 else 1.0

# Display Leakage-free Performance Scorecard
print("\n" + "="*55)
print("     LEAKAGE-FREE GROUP-SPLIT KNN SCORECARD      ")
print("="*55)
print(f"Training Time:                    {training_time_g:.4f} seconds")
print(f"Inference Time:                   {inference_time_g:.4f} seconds")
print(f"Accuracy:                         {accuracy_g:.5%}")
print(f"Precision:                        {precision_g:.5%}")
print(f"Recall:                           {recall_g:.5%}")
print(f"F1-Score:                         {f1_g:.5%}")
print(f"Observation-level Top-1 Accuracy: {top1_accuracy_g:.5%}")
print(f"Candidate Miss Rate:              {candidate_miss_rate_g:.5%}")
print(f"Route Continuity:                 {route_continuity_g:.5%}")
print("="*55)
print("\nConfusion Matrix:")
print(conf_matrix_g)
```
